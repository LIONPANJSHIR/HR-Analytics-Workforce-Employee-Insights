```python
files = {
    "sql": [
        "00_init_database.sql",
        "01_create_database.sql",
        "02_load_raw.sql",
        "03_data_quality.sql",
        "04_create_processed_tables.sql",
        "05_transform_processed.sql"
    ],
    "notebooks": ["01_data_audit.ipynb", "02_eda.ipynb"],
    "powerbi": ["hr_analytics.pbix"],
    "reports": ["figures/"],
    "docs": ["README.md", "requirements.txt", "data_dictionary.md", ".gitignore"],
}


root_dir = Path(os.getcwd())
for file_type, file_list in files.items():
    os.makedirs(root_dir / file_type, exist_ok=True)
    for file_name in file_list:
        file_path = root_dir / file_type / file_name
        if not file_path.exists():
            if file_name.endswith("/"):
                os.makedirs(file_path, exist_ok=True)
            else:
                file_path.touch()






files =["requirements.txt", "README.md", ".gitignore"]

for file in files :
    file_path = root_dir / file 
    if not file_path.exists():
        file_path.touch()
```
